```python
import os
from pathlib import Path

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import torchvision.transforms as T
import matplotlib.pyplot as plt
from tqdm.auto import tqdm, trange

# SwinIR
import sys
sys.path.append("SwinIR")
from models.network_swinir import SwinIR

# Identity backbone (PyTorch)
from facenet_pytorch import InceptionResnetV1

# ===== Basic config =====
data_root = "dataset_256"   # <---- CHANGE THIS
lr_size   = 64
hr_size   = 256
batch_size = 8
learning_rate = 1e-4
num_epochs = 50

device = torch.device("mps" if torch.backends.mps.is_available() else "cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

torch.set_num_threads(4)
```

```python
class PairedSRDataset(Dataset):
    """
    Paired LR/HR dataset for super-resolution.

    Expects:
    root/train/LR, root/train/HR
    root/val/LR,   root/val/HR
    root/test/LR,  root/test/HR
    """
    def __init__(self, root, split="train", lr_size=64, hr_size=256):
        self.root = Path(root)
        split_dir = self.root / split
        self.lr_dir = split_dir / "LR"
        self.hr_dir = split_dir / "HR"

        hr_paths = sorted(list(self.hr_dir.glob("*")))
        lr_paths = sorted(list(self.lr_dir.glob("*")))

        assert len(hr_paths) > 0, f"No HR images found in {split}/HR"
        assert len(lr_paths) > 0, f"No LR images found in {split}/LR"

        lr_map = {p.stem: p for p in lr_paths}
        self.pairs = []
        for hr_p in hr_paths:
            stem = hr_p.stem
            if stem in lr_map:
                self.pairs.append((lr_map[stem], hr_p))

        assert len(self.pairs) > 0, f"No LR/HR pairs matched for split={split}"

        self.lr_transform = T.Compose([
            T.Resize((lr_size, lr_size), interpolation=Image.BICUBIC),
            T.ToTensor(),
        ])
        self.hr_transform = T.Compose([
            T.Resize((hr_size, hr_size), interpolation=Image.BICUBIC),
            T.ToTensor(),
        ])

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, idx):
        lr_path, hr_path = self.pairs[idx]
        lr = Image.open(lr_path).convert("RGB")
        hr = Image.open(hr_path).convert("RGB")

        return self.lr_transform(lr), self.hr_transform(hr)
```

```python
train_dataset = PairedSRDataset(data_root, split="train", lr_size=lr_size, hr_size=hr_size)
val_dataset   = PairedSRDataset(data_root, split="val",   lr_size=lr_size, hr_size=hr_size)
test_dataset  = PairedSRDataset(data_root, split="test",  lr_size=lr_size, hr_size=hr_size)

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True,  num_workers=0, drop_last=True)
val_loader   = DataLoader(val_dataset,   batch_size=batch_size, shuffle=False, num_workers=0, drop_last=False)
test_loader  = DataLoader(test_dataset,  batch_size=batch_size, shuffle=False, num_workers=0, drop_last=False)

print(f"Train samples: {len(train_dataset)}")
print(f"Val samples:   {len(val_dataset)}")
print(f"Test samples:  {len(test_dataset)}")
```

```python
model = SwinIR(
    upscale=4,          # 64 → 256
    in_chans=3,
    img_size=lr_size,   # 64
    window_size=8,
    img_range=1.0,
    depths=[6, 6, 6, 6],
    embed_dim=60,
    num_heads=[6, 6, 6, 6],
    mlp_ratio=2.0,
    upsampler="pixelshuffle",
).to(device)

print("SwinIR params:", sum(p.numel() for p in model.parameters())/1e6, "M")
```

```python
import torchvision.models as models

class VGGPerceptualLoss(nn.Module):
    def __init__(self, layer_name='relu4_1', use_l1=True):
        super().__init__()
        vgg = models.vgg19(weights=models.VGG19_Weights.IMAGENET1K_V1).features
        self.use_l1 = use_l1

        # pick layers up to relu4_1 (~layer index 21)
        self.slice = nn.Sequential(*list(vgg.children())[:22])
        for p in self.slice.parameters():
            p.requires_grad = False
        self.slice.eval()

    def forward(self, x, y):
        # assume x,y in [0,1]
        x_v = self.slice(x)
        y_v = self.slice(y).detach()
        if self.use_l1:
            return F.l1_loss(x_v, y_v)
        else:
            return F.mse_loss(x_v, y_v)

vgg_loss_fn = VGGPerceptualLoss(layer_name='relu4_1', use_l1=True).to(device)
```

```python
# Facenet backbone (frozen weights)
id_backbone = InceptionResnetV1(pretrained='vggface2', classify=False).to(device)
for p in id_backbone.parameters():
    p.requires_grad = False
id_backbone.eval()

class FacenetIdentityLoss(nn.Module):
    def __init__(self, backbone):
        super().__init__()
        self.backbone = backbone

    def forward(self, sr, hr):
        """
        sr, hr: (N,3,H,W) in [0,1]
        Returns scalar identity loss = 1 - cos(mean embeddings)
        """
        # resize to 160x160 for Facenet
        sr_160 = F.interpolate(sr, size=(160,160), mode='bilinear', align_corners=False)
        hr_160 = F.interpolate(hr, size=(160,160), mode='bilinear', align_corners=False)

        # normalize to [-1,1] as Facenet expects
        sr_160 = (sr_160 - 0.5) * 2.0
        hr_160 = (hr_160 - 0.5) * 2.0

        # embeddings; grad flows only through sr
        emb_hr = self.backbone(hr_160).detach()              # no grad wrt hr
        emb_sr = self.backbone(sr_160)                       # grad wrt sr

        emb_hr = F.normalize(emb_hr, p=2, dim=1)
        emb_sr = F.normalize(emb_sr, p=2, dim=1)

        cos = (emb_sr * emb_hr).sum(dim=1)                   # (N,)
        return 1.0 - cos.mean()

id_loss_fn = FacenetIdentityLoss(id_backbone).to(device)
```

```python
l1_criterion = nn.L1Loss(reduction='mean')

lambda_perc = 0.1
lambda_id   = 0.5   # you can tune this

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=learning_rate,
    betas=(0.9, 0.99),
    weight_decay=1e-4,
)

scheduler = torch.optim.lr_scheduler.MultiStepLR(
    optimizer,
    milestones=[10, 15],
    gamma=0.5,
)
```

```python
def train_one_epoch(model, optimizer, train_loader, device):
    model.train()
    running = {"total": 0.0, "l1": 0.0, "vgg": 0.0, "id": 0.0}
    n_samples = 0

    pbar = tqdm(train_loader, desc="Training")
    for lr_batch, hr_batch in pbar:
        lr_batch = lr_batch.to(device)   # (B,3,64,64)
        hr_batch = hr_batch.to(device)   # (B,3,256,256)
        bsz = lr_batch.size(0)
        n_samples += bsz

        optimizer.zero_grad()

        sr_batch = model(lr_batch)       # (B,3,256,256)

        loss_l1  = l1_criterion(sr_batch, hr_batch)
        loss_vgg = vgg_loss_fn(sr_batch, hr_batch)
        loss_id  = id_loss_fn(sr_batch, hr_batch)

        loss_total = loss_l1 + lambda_perc * loss_vgg + lambda_id * loss_id
        loss_total.backward()
        optimizer.step()

        running["total"] += loss_total.item() * bsz
        running["l1"]    += loss_l1.item()      * bsz
        running["vgg"]   += loss_vgg.item()     * bsz
        running["id"]    += loss_id.item()      * bsz

        pbar.set_postfix({
            "total": f"{loss_total.item():.3f}",
            "L1":    f"{loss_l1.item():.3f}",
            "VGG":   f"{loss_vgg.item():.3f}",
            "ID":    f"{loss_id.item():.3f}",
        })

    avg = {k: v / n_samples for k, v in running.items()}
    return avg


@torch.no_grad()
def validate(model, val_loader, device):
    model.eval()
    running = {"total": 0.0, "l1": 0.0, "vgg": 0.0, "id": 0.0}
    n_samples = 0

    pbar = tqdm(val_loader, desc="Validation")
    for lr_batch, hr_batch in pbar:
        lr_batch = lr_batch.to(device)
        hr_batch = hr_batch.to(device)
        bsz = lr_batch.size(0)
        n_samples += bsz

        sr_batch = model(lr_batch)

        loss_l1  = l1_criterion(sr_batch, hr_batch)
        loss_vgg = vgg_loss_fn(sr_batch, hr_batch)
        loss_id  = id_loss_fn(sr_batch, hr_batch)

        loss_total = loss_l1 + lambda_perc * loss_vgg + lambda_id * loss_id

        running["total"] += loss_total.item() * bsz
        running["l1"]    += loss_l1.item()      * bsz
        running["vgg"]   += loss_vgg.item()     * bsz
        running["id"]    += loss_id.item()      * bsz

        pbar.set_postfix({
            "total": f"{loss_total.item():.3f}",
            "L1":    f"{loss_l1.item():.3f}",
            "VGG":   f"{loss_vgg.item():.3f}",
            "ID":    f"{loss_id.item():.3f}",
        })

    avg = {k: v / n_samples for k, v in running.items()}
    return avg
```

```python
CHECKPOINT_DIR   = "./checkpoints_64to256"
LATEST_CKPT_PATH = os.path.join(CHECKPOINT_DIR, "latest.pth")
BEST_CKPT_PATH   = os.path.join(CHECKPOINT_DIR, "best.pth")

RESUME_TRAINING  = True

os.makedirs(CHECKPOINT_DIR, exist_ok=True)

start_epoch = 1
best_val_total = float("inf")

# ---- 1. Initialize empty histories (will be overwritten if resuming) ----
train_history = {"total": [], "l1": [], "vgg": [], "id": []}
val_history   = {"total": [], "l1": [], "vgg": [], "id": []}

# ---- 2. Resume checkpoint (including histories if present) ----
if RESUME_TRAINING and os.path.exists(LATEST_CKPT_PATH):
    print(f"\nResuming from checkpoint: {LATEST_CKPT_PATH}")
    ckpt = torch.load(LATEST_CKPT_PATH, map_location=device)

    model.load_state_dict(ckpt["model_state"])
    optimizer.load_state_dict(ckpt["optimizer_state"])
    scheduler.load_state_dict(ckpt["scheduler_state"])

    start_epoch = ckpt["epoch"] + 1
    best_val_total = ckpt.get("best_val_total", float("inf"))

    # NEW: restore history if it exists
    if "train_history" in ckpt and "val_history" in ckpt:
        train_history = ckpt["train_history"]
        val_history   = ckpt["val_history"]

    print(f"Resumed from epoch {ckpt['epoch']}, next epoch = {start_epoch}")
else:
    print("\nNo checkpoint loaded, starting from scratch.")


# ---- 3. Training loop ----
for epoch in trange(start_epoch, num_epochs + 1, desc="Epochs"):
    train_stats = train_one_epoch(model, optimizer, train_loader, device)
    val_stats   = validate(model, val_loader, device)
    scheduler.step()

    # Append stats to history
    for k in train_history.keys():
        train_history[k].append(train_stats[k])
        val_history[k].append(val_stats[k])

    print(f"\nEpoch {epoch}/{num_epochs}")
    print(f"Train | total: {train_stats['total']:.4f}, "
          f"L1: {train_stats['l1']:.4f}, "
          f"VGG: {train_stats['vgg']:.4f}, "
          f"ID: {train_stats['id']:.4f}")
    print(f"Val   | total: {val_stats['total']:.4f}, "
          f"L1: {val_stats['l1']:.4f}, "
          f"VGG: {val_stats['vgg']:.4f}, "
          f"ID: {val_stats['id']:.4f}")

    # ---- 4. Build checkpoint dict including full history ----
    ckpt_latest = {
        "epoch": epoch,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
        "train_stats": train_stats,   # last epoch stats
        "val_stats": val_stats,       # last epoch stats
        "train_history": train_history,  # full curves
        "val_history": val_history,      # full curves
        "best_val_total": best_val_total,
    }
    torch.save(ckpt_latest, LATEST_CKPT_PATH)

    # ---- 5. Best checkpoint ----
    if val_stats["total"] < best_val_total:
        best_val_total = val_stats["total"]
        ckpt_best = {
            **ckpt_latest,
            "best_val_total": best_val_total,
        }
        torch.save(ckpt_best, BEST_CKPT_PATH)
        print("✅ New best model saved.")
```

```python
import math

def psnr_batch(x, y, max_val=1.0):
    """
    x, y: tensors of shape (N,3,H,W) in [0,1]
    returns mean PSNR over batch
    """
    mse = F.mse_loss(x, y, reduction="none")
    mse = mse.view(mse.size(0), -1).mean(dim=1)  # per-image MSE
    # avoid log(0)
    mse = torch.clamp(mse, min=1e-8)
    psnr = 10 * torch.log10((max_val ** 2) / mse)
    return psnr.mean().item()
```

```python
import os
import torch
import torch.nn.functional as F
from torchvision.utils import save_image
from tqdm import tqdm

@torch.no_grad()
def evaluate_sr(model, loader, device, hr_size=256, save_dir="eval_results"):
    model.eval()

    # Create output directory for images
    os.makedirs(save_dir, exist_ok=True)

    # Running sums for SR vs HR
    sr_metrics = {
        "l1": 0.0,
        "vgg": 0.0,
        "id": 0.0,
        "cos": 0.0,
        "psnr": 0.0,
    }

    # Running sums for baseline (LR↑ vs HR)
    lr_metrics = {
        "l1": 0.0,
        "vgg": 0.0,
        "id": 0.0,
        "cos": 0.0,
        "psnr": 0.0,
    }

    n_samples = 0
    global_idx = 0  # for naming saved images

    pbar = tqdm(loader, desc="Evaluating (test)")
    for lr_batch, hr_batch in pbar:
        lr_batch = lr_batch.to(device)
        hr_batch = hr_batch.to(device)
        bsz = lr_batch.size(0)
        n_samples += bsz

        # SR prediction
        sr_batch = model(lr_batch)

        # Baseline: nearest upscale LR to HR size
        lr_up_batch = F.interpolate(
            lr_batch,
            size=hr_batch.shape[-2:],
            mode="nearest",
        )

        # ---- Compute metrics (baseline) ----
        l1_lr  = l1_criterion(lr_up_batch, hr_batch).item()
        vgg_lr = vgg_loss_fn(lr_up_batch, hr_batch).item()
        id_lr  = id_loss_fn(lr_up_batch, hr_batch).item()
        cos_lr = 1.0 - id_lr
        psnr_lr = psnr_batch(lr_up_batch, hr_batch)

        lr_metrics["l1"]   += l1_lr  * bsz
        lr_metrics["vgg"]  += vgg_lr * bsz
        lr_metrics["id"]   += id_lr  * bsz
        lr_metrics["cos"]  += cos_lr * bsz
        lr_metrics["psnr"] += psnr_lr * bsz

        # ---- Compute metrics (SR) ----
        l1_sr  = l1_criterion(sr_batch, hr_batch).item()
        vgg_sr = vgg_loss_fn(sr_batch, hr_batch).item()
        id_sr  = id_loss_fn(sr_batch, hr_batch).item()
        cos_sr = 1.0 - id_sr
        psnr_sr = psnr_batch(sr_batch, hr_batch)

        sr_metrics["l1"]   += l1_sr  * bsz
        sr_metrics["vgg"]  += vgg_sr * bsz
        sr_metrics["id"]   += id_sr  * bsz
        sr_metrics["cos"]  += cos_sr * bsz
        sr_metrics["psnr"] += psnr_sr * bsz

        pbar.set_postfix({
            "SR_L1": f"{l1_sr:.3f}",
            "SR_ID": f"{id_sr:.3f}",
            "SR_PSNR": f"{psnr_sr:.2f}",
        })

        # ---- Save LR↑, SR, HR images ----
        for i in range(bsz):
            lr_path = os.path.join(save_dir, f"sample_{global_idx:05d}_lr.png")
            sr_path = os.path.join(save_dir, f"sample_{global_idx:05d}_sr.png")
            hr_path = os.path.join(save_dir, f"sample_{global_idx:05d}_hr.png")

            save_image(lr_up_batch[i].cpu(), lr_path)
            save_image(sr_batch[i].cpu(), sr_path)
            save_image(hr_batch[i].cpu(), hr_path)

            global_idx += 1

    # ---- Average metrics ----
    for k in sr_metrics:
        sr_metrics[k] /= n_samples
        lr_metrics[k] /= n_samples

    print(f"Saved {global_idx} LR/SR/HR images to: {save_dir}")

    return lr_metrics, sr_metrics
```

```python
import torch
import torch.nn.functional as F
import numpy as np

@torch.no_grad()
def get_normalized_embeddings(x, backbone, device):
    """
    x: (N,3,H,W) in [0,1] torch tensor
    backbone: InceptionResnetV1 model
    returns: (N,D) numpy array with L2-normalized embeddings
    """
    x = x.to(device)
    # Resize to 160x160 as in training
    x160 = F.interpolate(x, size=(160,160), mode='bilinear', align_corners=False)
    # Normalize to [-1,1]
    x160 = (x160 - 0.5) * 2.0
    # Embeddings
    emb = backbone(x160)            # (N,D)
    emb = F.normalize(emb, p=2, dim=1)
    return emb.cpu().numpy()
```

```python
@torch.no_grad()
def collect_test_embeddings(model, backbone, test_loader, device):
    model.eval()
    all_emb_lr = []
    all_emb_sr = []
    all_emb_hr = []

    for lr_batch, hr_batch in tqdm(test_loader, desc="Collecting embeddings"):
        lr_batch = lr_batch.to(device)  # (B,3,64,64)
        hr_batch = hr_batch.to(device)  # (B,3,256,256)

        # SR
        sr_batch = model(lr_batch)      # (B,3,256,256)

        # Baseline: upsample LR to HR size (MPS-safe)
        lr_up_batch = F.interpolate(
            lr_batch,
            size=hr_batch.shape[-2:],   # (256,256)
            mode="bilinear",
            align_corners=False,
        )

        # Get embeddings (N,D)
        emb_lr = get_normalized_embeddings(lr_up_batch, backbone, device)
        emb_sr = get_normalized_embeddings(sr_batch,    backbone, device)
        emb_hr = get_normalized_embeddings(hr_batch,    backbone, device)

        all_emb_lr.append(emb_lr)
        all_emb_sr.append(emb_sr)
        all_emb_hr.append(emb_hr)

    emb_lr_up = np.vstack(all_emb_lr)   # (N,D)
    emb_sr    = np.vstack(all_emb_sr)   # (N,D)
    emb_hr    = np.vstack(all_emb_hr)   # (N,D)
    return emb_lr_up, emb_sr, emb_hr
```

```python
def build_genuine_impostor_scores(emb_x, emb_hr):
    """
    emb_x : (N,D) normalized embeddings (LR↑ or SR)
    emb_hr: (N,D) normalized HR embeddings (same order)
    Returns:
        genuine_scores: (N,)
        impostor_scores: (N*(N-1),)
    """
    # Cosine similarity = dot product since embeddings are normalized
    sim_matrix = emb_x @ emb_hr.T   # (N,N)

    N = sim_matrix.shape[0]
    # Genuine scores = diagonal
    genuine_scores = np.diag(sim_matrix)

    # Impostor = all off-diagonal elements
    mask = ~np.eye(N, dtype=bool)
    impostor_scores = sim_matrix[mask]

    return genuine_scores, impostor_scores
```

```python
gen_lr, imp_lr = build_genuine_impostor_scores(emb_lr_up, emb_hr)
gen_sr, imp_sr = build_genuine_impostor_scores(emb_sr,    emb_hr)
```

```python
from sklearn.metrics import roc_curve

def build_labels_and_scores(genuine, impostor):
    """
    genuine: (Ng,)
    impostor: (Ni,)
    Returns:
        scores: (Ng+Ni,)
        labels: same shape, 1 for genuine, 0 for impostor
    """
    genuine = np.asarray(genuine)
    impostor = np.asarray(impostor)

    scores = np.concatenate([genuine, impostor], axis=0)
    labels = np.concatenate([np.ones_like(genuine), np.zeros_like(impostor)], axis=0)
    return scores, labels


def compute_far_at_tar(scores, labels, target_tar=0.95):
    """
    scores: similarity scores (higher = more similar)
    labels: 1 for genuine, 0 for impostor
    target_tar: desired True Acceptance Rate (e.g. 0.95, 0.99)

    Returns:
        far_at_tar, chosen_threshold, actual_tar
    """
    fpr, tpr, thresholds = roc_curve(labels, scores)
    # TAR = TPR
    tar = tpr
    # Find index closest to desired TAR
    idx = np.argmin(np.abs(tar - target_tar))
    far_at_tar = fpr[idx]
    thr = thresholds[idx]
    actual_tar = tar[idx]
    return far_at_tar, thr, actual_tar
```

```python
scores_lr, labels_lr = build_labels_and_scores(gen_lr, imp_lr)
scores_sr, labels_sr = build_labels_and_scores(gen_sr, imp_sr)
```

```python
for target_tar in [0.90, 0.95, 0.99]:
    far_lr, thr_lr, tar_lr = compute_far_at_tar(scores_lr, labels_lr, target_tar)
    far_sr, thr_sr, tar_sr = compute_far_at_tar(scores_sr, labels_sr, target_tar)

    print(f"\n=== FAR@TAR={target_tar:.2f} ===")
    print(f"Baseline (LR↑): FAR = {far_lr:.4f}, TAR = {tar_lr:.4f}, thr = {thr_lr:.3f}")
    print(f"Ours (SR)    : FAR = {far_sr:.4f}, TAR = {tar_sr:.4f}, thr = {thr_sr:.3f}")
    print(f"ΔFAR = {far_lr - far_sr:+.4f}")
```

```python
# Example: if you already computed these in your other eval:
# lr_metrics = {"l1": ..., "vgg": ..., "id": ..., "cos": ..., "psnr": ...}
# sr_metrics = { ... }

def print_comparison_table(lr_metrics, sr_metrics, scores_lr, labels_lr, scores_sr, labels_sr):
    # Compute FAR@TAR at a couple of operating points
    tars = [0.90, 0.95, 0.99]
    far_tar_rows = []
    for target_tar in tars:
        far_lr, _, _ = compute_far_at_tar(scores_lr, labels_lr, target_tar)
        far_sr, _, _ = compute_far_at_tar(scores_sr, labels_sr, target_tar)
        far_tar_rows.append((target_tar, far_lr, far_sr))

    print("\n===== Test Set Summary (Baseline LR↑ vs Ours SR) =====")
    print(f"{'Metric':<12} {'LR↑':>12} {'SR':>12} {'Δ (LR↑-SR)':>14}")
    print("-"*54)
    for k in ["l1", "vgg", "id", "cos", "psnr"]:
        v_lr = lr_metrics[k]
        v_sr = sr_metrics[k]
        print(f"{k.upper():<12} {v_lr:>12.4f} {v_sr:>12.4f} {v_lr - v_sr:>+14.4f}")

    print("\nFAR@TAR comparison:")
    print(f"{'TAR':<8} {'FAR_LR↑':>12} {'FAR_SR':>12} {'ΔFAR':>12}")
    print("-"*46)
    for tar, far_lr, far_sr in far_tar_rows:
        print(f"{tar:<8.2f} {far_lr:>12.5f} {far_sr:>12.5f} {far_lr - far_sr:>+12.5f}")
```

```python
import torch
import torch.nn.functional as F
from torchvision import transforms
from PIL import Image
import matplotlib.pyplot as plt
import os

# ---- Inference function ----
def super_resolve_image(
    img_path,
    model,
    device,
    save_path="sr_output.png",
    save_lr_path="lr_64x64.png",
    upscale=4
):
    model.eval()

    # ---- Load + resize to 64x64 ----
    img = Image.open(img_path).convert("RGB")
    lr_img = img.resize((64, 64), Image.BICUBIC)   # <-- resize

    # Save resized LR
    lr_img.save(save_lr_path)
    print(f"Resized LR image saved to {save_lr_path}")

    # ---- Transform to tensor ----
    transform = transforms.Compose([
        transforms.ToTensor(),      # [0,1]
    ])
    lr_tensor = transform(lr_img).unsqueeze(0).to(device)   # (1,3,64,64)

    with torch.no_grad():
        # Run SR model
        sr_tensor = model(lr_tensor)   # (1,3,256,256)

    # ---- Convert for display ----
    lr_vis = lr_tensor[0].cpu().permute(1,2,0).numpy()
    sr_vis = sr_tensor[0].cpu().permute(1,2,0).numpy()

    # ---- Save SR image ----
    sr_img = transforms.ToPILImage()(sr_tensor.squeeze(0).cpu())
    sr_img.save(save_path)
    print(f"SR image saved to {save_path}")

    # ---- Display LR and SR ----
    plt.figure(figsize=(10,4))
    plt.subplot(1,2,1)
    plt.imshow(lr_vis)
    plt.title("Low-Resolution Input (64×64)")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.imshow(sr_vis)
    plt.title("Super-Resolved Output")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    return lr_img, sr_img
```

```python
import cv2
import numpy as np

def add_cctv_noise(
    input_path: str,
    output_path: str = "cctv_noisy.png",
    downscale_factor: float = 0.25,
    jpeg_quality: int = 30,
    noise_std: float = 15.0,
    motion_blur_ksize: int = 7,
):
    """
    Load an image, apply CCTV-like degradations, and save it.

    Args:
        input_path:  path to the clean input image.
        output_path: where to save the degraded image.
        downscale_factor: how much to shrink the image before upsampling (0.25 = 4x smaller).
        jpeg_quality: lower = more blocky artifacts (typical CCTV: 10–40).
        noise_std: standard deviation of added Gaussian noise (0–30 is reasonable).
        motion_blur_ksize: kernel size for motion blur (odd number, e.g. 5, 7, 9).
    """
    # --- 1. Read image ---
    img_bgr = cv2.imread(input_path, cv2.IMREAD_COLOR)
    if img_bgr is None:
        raise ValueError(f"Could not read image from '{input_path}'")
    img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB).astype(np.float32)

    h, w = img.shape[:2]

    # --- 2. Downscale + upsample (low resolution) ---
    new_w = max(1, int(w * downscale_factor))
    new_h = max(1, int(h * downscale_factor))

    small = cv2.resize(img, (new_w, new_h), interpolation=cv2.INTER_AREA)
    # Upsample with NEAREST to get blockiness
    lowres = cv2.resize(small, (w, h), interpolation=cv2.INTER_NEAREST)

    # --- 3. Add Gaussian noise ---
    if noise_std > 0:
        noise = np.random.normal(0, noise_std, size=lowres.shape).astype(np.float32)
        noisy = lowres + noise
    else:
        noisy = lowres

    # Clip to valid range
    noisy = np.clip(noisy, 0, 255)

    # --- 4. Motion blur (simulating camera jitter) ---
    if motion_blur_ksize > 1 and motion_blur_ksize % 2 == 1:
        # horizontal motion blur kernel
        kernel = np.zeros((motion_blur_ksize, motion_blur_ksize), dtype=np.float32)
        kernel[motion_blur_ksize // 2, :] = 1.0 / motion_blur_ksize
        blurred = cv2.filter2D(noisy, -1, kernel)
    else:
        blurred = noisy

    # --- 5. Fake CCTV compression (JPEG) ---
    # Convert back to BGR for OpenCV JPEG encode
    blurred_bgr = cv2.cvtColor(blurred.astype(np.uint8), cv2.COLOR_RGB2BGR)

    encode_param = [int(cv2.IMWRITE_JPEG_QUALITY), int(jpeg_quality)]
    result, encimg = cv2.imencode(".jpg", blurred_bgr, encode_param)
    if not result:
        raise RuntimeError("JPEG encoding failed")

    # Decode back to get block artifacts
    cctv_bgr = cv2.imdecode(encimg, cv2.IMREAD_COLOR)

    # --- 6. Save final result ---
    cv2.imwrite(output_path, cctv_bgr)
    print(f"CCTV-style noisy image saved to: {output_path}")
```
