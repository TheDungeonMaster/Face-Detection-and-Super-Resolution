```python
import os
from pathlib import Path

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import torchvision.transforms as T
import matplotlib.pyplot as plt
from tqdm.auto import tqdm, trange

# SwinIR
import sys
sys.path.append("SwinIR")
from models.network_swinir import SwinIR

# Identity backbone (PyTorch)
from facenet_pytorch import InceptionResnetV1

# ===== Basic config =====
data_root = "dataset_256"   # <---- CHANGE THIS
lr_size   = 64
hr_size   = 256
batch_size = 8
learning_rate = 1e-4
num_epochs = 30

device = torch.device("mps" if torch.backends.mps.is_available() else "cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

torch.set_num_threads(4)
```

```python
class PairedSRDataset(Dataset):
    """
    Paired LR/HR dataset for super-resolution.

    Expects:
    root/train/LR, root/train/HR
    root/val/LR,   root/val/HR
    root/test/LR,  root/test/HR
    """
    def __init__(self, root, split="train", lr_size=64, hr_size=256):
        self.root = Path(root)
        split_dir = self.root / split
        self.lr_dir = split_dir / "LR"
        self.hr_dir = split_dir / "HR"

        hr_paths = sorted(list(self.hr_dir.glob("*")))
        lr_paths = sorted(list(self.lr_dir.glob("*")))

        assert len(hr_paths) > 0, f"No HR images found in {split}/HR"
        assert len(lr_paths) > 0, f"No LR images found in {split}/LR"

        lr_map = {p.stem: p for p in lr_paths}
        self.pairs = []
        for hr_p in hr_paths:
            stem = hr_p.stem
            if stem in lr_map:
                self.pairs.append((lr_map[stem], hr_p))

        assert len(self.pairs) > 0, f"No LR/HR pairs matched for split={split}"

        self.lr_transform = T.Compose([
            T.Resize((lr_size, lr_size), interpolation=Image.BICUBIC),
            T.ToTensor(),
        ])
        self.hr_transform = T.Compose([
            T.Resize((hr_size, hr_size), interpolation=Image.BICUBIC),
            T.ToTensor(),
        ])

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, idx):
        lr_path, hr_path = self.pairs[idx]
        lr = Image.open(lr_path).convert("RGB")
        hr = Image.open(hr_path).convert("RGB")

        return self.lr_transform(lr), self.hr_transform(hr)
```

```python
train_dataset = PairedSRDataset(data_root, split="train", lr_size=lr_size, hr_size=hr_size)
val_dataset   = PairedSRDataset(data_root, split="val",   lr_size=lr_size, hr_size=hr_size)
test_dataset  = PairedSRDataset(data_root, split="test",  lr_size=lr_size, hr_size=hr_size)

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True,  num_workers=0, drop_last=True)
val_loader   = DataLoader(val_dataset,   batch_size=batch_size, shuffle=False, num_workers=0, drop_last=False)
test_loader  = DataLoader(test_dataset,  batch_size=batch_size, shuffle=False, num_workers=0, drop_last=False)

print(f"Train samples: {len(train_dataset)}")
print(f"Val samples:   {len(val_dataset)}")
print(f"Test samples:  {len(test_dataset)}")
```

```python
model = SwinIR(
    upscale=4,
    in_chans=3,
    img_size=lr_size,   # 64
    window_size=8,
    img_range=1.0,
    depths=[6, 6, 6, 6],
    embed_dim=60,
    num_heads=[6, 6, 6, 6],
    mlp_ratio=2.0,
    upsampler="pixelshuffle",
).to(device)

print("Model params:", sum(p.numel() for p in model.parameters())/1e6, "M")
```

```python
import torch
import torch.nn as nn
from torchvision.models import vgg19, VGG19_Weights


class VGGPerceptualLoss(nn.Module):
    def __init__(self, layer_name='relu4_1', use_l1=True):
        super().__init__()

        # Load pretrained VGG19
        vgg = vgg19(weights=VGG19_Weights.IMAGENET1K_V1).features

        # Map human-readable names to index in vgg.features
        # VGG19: conv/relu layers order; we pick until relu4_1 (~layer 21)
        layer_map = {
            'relu1_1': 1,
            'relu1_2': 3,
            'relu2_1': 6,
            'relu2_2': 8,
            'relu3_1': 11,
            'relu3_2': 13,
            'relu3_3': 15,
            'relu3_4': 17,
            'relu4_1': 20,
            'relu4_2': 22,
            'relu4_3': 24,
            'relu4_4': 26,
        }
        assert layer_name in layer_map, f"Unknown layer_name: {layer_name}"
        cutoff = layer_map[layer_name]

        # Keep VGG up to the chosen layer
        self.vgg = nn.Sequential(*[vgg[i] for i in range(cutoff + 1)])
        for p in self.vgg.parameters():
            p.requires_grad = False  # freeze

        # ImageNet normalization parameters
        self.register_buffer('mean', torch.tensor([0.485, 0.456, 0.406]).view(1,3,1,1))
        self.register_buffer('std',  torch.tensor([0.229, 0.224, 0.225]).view(1,3,1,1))

        # Distance in feature space
        self.criterion = nn.L1Loss() if use_l1 else nn.MSELoss()

    def normalize(self, x):
        # Input x is assumed to be in [0, 1]
        return (x - self.mean) / self.std

    def forward(self, sr, hr):
        """
        sr, hr: (N,3,H,W) in [0,1]
        """
        sr_norm = self.normalize(sr)
        hr_norm = self.normalize(hr)

        with torch.no_grad():
            feat_hr = self.vgg(hr_norm)
        # Allow gradients only through SR side
        feat_sr = self.vgg(sr_norm)

        loss = self.criterion(feat_sr, feat_hr)
        return loss

vgg_loss_fn = VGGPerceptualLoss().to(device)
```

```python
id_backbone = InceptionResnetV1(pretrained='vggface2', classify=False).to(device)
id_backbone.eval()
for p in id_backbone.parameters():
    p.requires_grad = False

class FacenetIdentityLoss(nn.Module):
    def __init__(self, backbone):
        super().__init__()
        self.backbone = backbone

    def forward(self, sr, hr):
        # Resize to 160x160
        sr_160 = F.interpolate(sr, size=(160,160), mode='bilinear', align_corners=False)
        hr_160 = F.interpolate(hr, size=(160,160), mode=( 'bilinear' ), align_corners=False)

        # Normalize to [-1, 1]
        sr_160 = (sr_160 - 0.5) * 2.0
        hr_160 = (hr_160 - 0.5) * 2.0

        # Extract embeddings
        emb_sr = self.backbone(sr_160)              # gradients flow through SR
        emb_hr = self.backbone(hr_160).detach()     # HR no grad

        # Normalize
        emb_sr = F.normalize(emb_sr, p=2, dim=1)
        emb_hr = F.normalize(emb_hr, p=2, dim=1)

        return 1.0 - (emb_sr * emb_hr).sum(dim=1).mean()

id_loss_fn = FacenetIdentityLoss(id_backbone).to(device)
```

```python
l1_criterion = nn.L1Loss()

lambda_perc = 0.1
lambda_id   = 0.5  # tune if needed

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=learning_rate,
    betas=(0.9, 0.99),
    weight_decay=1e-4,
)

scheduler = torch.optim.lr_scheduler.MultiStepLR(
    optimizer,
    milestones=[10, 15],
    gamma=0.5,
)
```

```python
def train_one_epoch(model, optimizer, loader):
    model.train()
    running = {"total":0, "l1":0, "vgg":0, "id":0}
    n_samples = 0

    pbar = tqdm(loader, desc="Training")
    for lr_batch, hr_batch in pbar:
        lr_batch = lr_batch.to(device)
        hr_batch = hr_batch.to(device)
        bsz = lr_batch.size(0)
        n_samples += bsz

        optimizer.zero_grad()

        sr = model(lr_batch)

        loss_l1  = l1_criterion(sr, hr_batch)
        loss_vgg = vgg_loss_fn(sr, hr_batch)
        loss_id  = id_loss_fn(sr, hr_batch)

        loss_total = loss_l1 + lambda_perc * loss_vgg + lambda_id * loss_id
        loss_total.backward()
        optimizer.step()

        running["total"] += loss_total.item() * bsz
        running["l1"]    += loss_l1.item() * bsz
        running["vgg"]   += loss_vgg.item() * bsz
        running["id"]    += loss_id.item() * bsz

        pbar.set_postfix({
            "total": loss_total.item(),
            "L1": loss_l1.item(),
            "VGG": loss_vgg.item(),
            "ID": loss_id.item(),
        })

    return {k: v/n_samples for k,v in running.items()}


@torch.no_grad()
def validate(model, loader):
    model.eval()
    running = {"total":0, "l1":0, "vgg":0, "id":0}
    n_samples = 0

    pbar = tqdm(loader, desc="Validation")
    for lr_batch, hr_batch in pbar:
        lr_batch = lr_batch.to(device)
        hr_batch = hr_batch.to(device)
        bsz = lr_batch.size(0)
        n_samples += bsz

        sr = model(lr_batch)

        loss_l1  = l1_criterion(sr, hr_batch)
        loss_vgg = vgg_loss_fn(sr, hr_batch)
        loss_id  = id_loss_fn(sr, hr_batch)

        loss_total = loss_l1 + lambda_perc * loss_vgg + lambda_id * loss_id

        running["total"] += loss_total.item() * bsz
        running["l1"]    += loss_l1.item() * bsz
        running["vgg"]   += loss_vgg.item() * bsz
        running["id"]    += loss_id.item() * bsz

        pbar.set_postfix({
            "total": loss_total.item(),
            "L1": loss_l1.item(),
            "VGG": loss_vgg.item(),
            "ID": loss_id.item(),
        })

    return {k: v/n_samples for k,v in running.items()}
```

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

class ArcFaceIdentityMetric(nn.Module):
    """
    Identity *metric* using insightface app.get(...).
    NOT differentiable, so don't use it as a training loss.
    """

    def __init__(self, app, device):
        super().__init__()
        self.app = app
        self.device = device

    def forward(self, sr, hr):
        """
        sr, hr: (N,3,H,W) in [0,1] torch tensors.
        Returns: mean identity loss = 1 - cosine_similarity(sr, hr).
        """
        sr_np = sr.detach().cpu().numpy()
        hr_np = hr.detach().cpu().numpy()

        sr_embs = []
        hr_embs = []

        for s, h in zip(sr_np, hr_np):
            # CHW -> HWC, [0,1] -> uint8
            sr_img_np = (np.transpose(s, (1, 2, 0)) * 255.0).clip(0, 255).astype(np.uint8)
            hr_img_np = (np.transpose(h, (1, 2, 0)) * 255.0).clip(0, 255).astype(np.uint8)

            # Get faces from insightface app
            sr_faces = self.app.get(sr_img_np)
            hr_faces = self.app.get(hr_img_np)

            if len(sr_faces) == 0 or len(hr_faces) == 0:
                # no detected face → skip this pair
                continue

            # Take first face; use normed_embedding if available, else embedding
            sr_face = sr_faces[0]
            hr_face = hr_faces[0]

            sr_emb = getattr(sr_face, "normed_embedding", sr_face.embedding)
            hr_emb = getattr(hr_face, "normed_embedding", hr_face.embedding)

            sr_embs.append(torch.from_numpy(sr_emb).float())
            hr_embs.append(torch.from_numpy(hr_emb).float())

        if len(sr_embs) == 0:
            # No faces in the whole batch – return 0 so it doesn't blow up your logs
            return torch.tensor(0.0, device=self.device)

        sr_embs = torch.stack(sr_embs, dim=0).to(self.device)
        hr_embs = torch.stack(hr_embs, dim=0).to(self.device)

        sr_embs = F.normalize(sr_embs, p=2, dim=1)
        hr_embs = F.normalize(hr_embs, p=2, dim=1)

        cos = F.cosine_similarity(sr_embs, hr_embs, dim=1)  # (M,)
        id_loss = 1.0 - cos.mean()
        return id_loss
```

```python
def train_one_epoch(model, optimizer, train_loader, device):
    model.train()
    running = {"total": 0.0, "l1": 0.0, "vgg": 0.0, "id": 0.0}
    n_samples = 0

    pbar = tqdm(train_loader, desc="Training")

    for lr_batch, hr_batch in pbar:
        lr_batch = lr_batch.to(device)   # (B,3,64,64)
        hr_batch = hr_batch.to(device)   # (B,3,256,256)
        bsz = lr_batch.size(0)
        n_samples += bsz

        optimizer.zero_grad()

        # --- SR forward ---
        sr_batch = model(lr_batch)       # (B,3,256,256)

        # --- Losses ---
        loss_l1  = l1_criterion(sr_batch, hr_batch)
        loss_vgg = vgg_loss_fn(sr_batch, hr_batch)

        # Resize for ArcFace (expects 112x112)
        # Use InsightFace detector
        sr_faces = app.get(sr_img_np)["embedding"]
        hr_faces = app.get(hr_img_np)["embedding"]

        loss_id = id_loss_fn(sr_face, hr_face)

        loss_total = loss_l1 + lambda_perc * loss_vgg + lambda_id * loss_id
        loss_total.backward()
        optimizer.step()

        running["total"] += loss_total.item() * bsz
        running["l1"]    += loss_l1.item()      * bsz
        running["vgg"]   += loss_vgg.item()     * bsz
        running["id"]    += loss_id.item()      * bsz

        pbar.set_postfix({
            "total": f"{loss_total.item():.3f}",
            "L1":    f"{loss_l1.item():.3f}",
            "VGG":   f"{loss_vgg.item():.3f}",
            "ID":    f"{loss_id.item():.3f}",
        })

    avg = {k: v / n_samples for k, v in running.items()}
    return avg


@torch.no_grad()
def validate(model, val_loader, device):
    model.eval()
    running = {"total": 0.0, "l1": 0.0, "vgg": 0.0, "id": 0.0}
    n_samples = 0

    pbar = tqdm(val_loader, desc="Validation")

    for lr_batch, hr_batch in pbar:
        lr_batch = lr_batch.to(device)
        hr_batch = hr_batch.to(device)
        bsz = lr_batch.size(0)
        n_samples += bsz

        sr_batch = model(lr_batch)

        loss_l1  = l1_criterion(sr_batch, hr_batch)
        loss_vgg = vgg_loss_fn(sr_batch, hr_batch)

        # Use InsightFace detector
        sr_faces = app.get(sr_img_np)["embedding"]
        hr_faces = app.get(hr_img_np)["embedding"]

        loss_id = id_loss_fn(sr_face, hr_face)

        loss_total = loss_l1 + lambda_perc * loss_vgg + lambda_id * loss_id

        running["total"] += loss_total.item() * bsz
        running["l1"]    += loss_l1.item()      * bsz
        running["vgg"]   += loss_vgg.item()     * bsz
        running["id"]    += loss_id.item()      * bsz

        pbar.set_postfix({
            "total": f"{loss_total.item():.3f}",
            "L1":    f"{loss_l1.item():.3f}",
            "VGG":   f"{loss_vgg.item():.3f}",
            "ID":    f"{loss_id.item():.3f}",
        })

    avg = {k: v / n_samples for k, v in running.items()}
    return avg
```

```python
CHECKPOINT_DIR = "./checkpoints_64to256"
LATEST_CKPT_PATH = f"{CHECKPOINT_DIR}/latest.pth"
BEST_CKPT_PATH   = f"{CHECKPOINT_DIR}/best.pth"

os.makedirs(CHECKPOINT_DIR, exist_ok=True)

start_epoch = 1
best_val_total = float("inf")

# ====== RESUME CHECKPOINT ======
if os.path.exists(LATEST_CKPT_PATH):
    print(f"🔄 Loading checkpoint: {LATEST_CKPT_PATH}")
    ckpt = torch.load(LATEST_CKPT_PATH, map_location=device)

    model.load_state_dict(ckpt["model_state"])
    optimizer.load_state_dict(ckpt["optimizer_state"])
    scheduler.load_state_dict(ckpt["scheduler_state"])

    start_epoch = ckpt["epoch"] + 1
    best_val_total = ckpt["val_total"]

    print(f"➡ Resumed at epoch {start_epoch}, best val so far = {best_val_total:.4f}")
else:
    print("🆕 No checkpoint found — starting from scratch.")


for epoch in trange(start_epoch, num_epochs + 1, desc="Epochs"):
    train_stats = train_one_epoch(model, optimizer, train_loader)
    val_stats   = validate(model, val_loader)
    scheduler.step()

    print(f"\nEpoch {epoch}/{num_epochs}")
    print("Train:", train_stats)
    print("Val:  ", val_stats)

    ckpt = {
        "epoch": epoch,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
        "val_total": val_stats["total"],
    }

    torch.save(ckpt, LATEST_CKPT_PATH)

    if val_stats["total"] < best_val_total:
        best_val_total = val_stats["total"]
        torch.save(ckpt, BEST_CKPT_PATH)
        print("✅ New BEST checkpoint saved.")
```
